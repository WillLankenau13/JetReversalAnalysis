```python
import pandas as pd
import numpy as np
import ast
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("data/reversals_dataset_40-test.csv")
```

```python
u_list = data['u_list'].tolist()
u_arrays = [ast.literal_eval(x) for x in u_list]
u_arrays = np.array(u_arrays)

psi_e_list = data['psi_e'].tolist()
psi_e_arrays = [ast.literal_eval(x) for x in psi_e_list]
psi_e_arrays = np.array(psi_e_arrays)

b_e_list = data['b_e'].tolist()
b_e_arrays = [ast.literal_eval(x) for x in b_e_list]
b_e_arrays = np.array(b_e_arrays)

```

```python
###Flip the reversal

for i in range(len(u_arrays)):
    u_arrays[i] = [-1 * x for x in u_arrays[i]]
    psi_e_arrays[i] = [-1 * x for x in psi_e_arrays[i]]
    b_e_arrays[i] = [-1 * x for x in b_e_arrays[i]]

```

```python
data['u_list'] = [str(x.tolist()) for x in u_arrays]
data['psi_e'] = [str(x.tolist()) for x in psi_e_arrays]
data['b_e'] = [str(x.tolist()) for x in b_e_arrays]
```

```python
data.to_csv('data/reversals_dataset_40_neg-test.csv', index=False)
```
